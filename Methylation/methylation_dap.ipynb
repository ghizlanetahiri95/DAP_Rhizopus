```python
# Import libraries 
import os
import pandas as pd
import numpy as np
```

```python

# Cargar el archivo de expresión
expresion = pd.read_csv("GeneID_FPKM_TPM_MACs_Luz.csv", sep=";")

# Definir el directorio donde se encuentran los archivos
directory = "C:/Users/34698/OneDrive/Escritorio/Trabajos_2024/Factores de transcripcion/DAP/Picos/Gene_id/"

# Obtener la lista de archivos en el directorio que contienen "peak_centric" en su nombre
blast2go_files = [f for f in os.listdir(directory) if "peak_centric.txtgene_id" in f]

# Iterar sobre cada archivo blast2go
for filename in blast2go_files:
    file_path = os.path.join(directory, filename)
    
    # Leer el archivo de blast2go
    df = pd.read_csv(file_path, sep="\t", header=None)
    df.columns = ["Ids"]  # Cambiar el nombre de la columna a "Ids"
    
    # Hacer merge con el archivo de expresión
    merge_df = pd.merge(df, expresion, left_on="Ids", right_on="GeneID")
    
    # Generar el nombre del archivo de salida
    output_filename = filename.replace("peak_centric", "expression")  # Reemplazar "peak_centric" por "expression"
    output_path = os.path.join(directory, output_filename)
    
    # Guardar el DataFrame filtrado en un nuevo archivo
    merge_df.to_csv(output_path, sep="\t", index=False, header=True)  # Incluye nombres de columnas
    
    print(f"Archivo filtrado guardado: {output_path}")
```

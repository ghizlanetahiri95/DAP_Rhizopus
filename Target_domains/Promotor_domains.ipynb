The following code was developed in order to study the domain organization of the gene targets of the transciption factors in the DAP analysis. 

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
```

```python
# Leer los datos en un DataFrame
data = pd.read_csv('Ghiz_plot_Data.tx', sep='\t')
data["ID"].nunique()
# Agrupar los datos por 'ID' para visualización
grouped = data.groupby('ID')
#grouped["motif"]=grouped["motif"].astype(str)
# Preparar el gráfico
plt.figure(figsize=(15, 10))

# Variable para manejar la posición en el eje Y
y_pos = 0
yticks = []
legend_labels = set()  # Usar un conjunto para evitar duplicados en la leyenda

# Graficar dominios para cada proteína
for seq_id, group in grouped:
    for idx, row in group.iterrows():
        # Añadir el tipo de dominio a la leyenda solo si no ha sido agregado previamente
        label = row['motif'] if row['motif'] not in legend_labels else None
        plt.plot([row['Start'], row['End']], [y_pos, y_pos], marker='o', label=label)
        if label:
            legend_labels.add(row['motif'])  # Marcar este dominio como agregado a la leyenda

    # Agregar el nombre de la proteína a la lista de etiquetas del eje Y
    yticks.append(seq_id)
    y_pos += 1

# Configurar el gráfico
plt.xlabel('Posición en la proteína')
plt.ylabel('Protein (ID)')
plt.yticks(range(len(yticks)), yticks)  # Asignar los nombres de proteínas al eje Y
plt.title('Distribución de dominios en las proteínas')
plt.grid(True)
# Mostrar el gráfico

plt.show()
```

```python
# Leer los datos en un DataFrame
data = pd.read_csv('Ghiz_plot_Data.txt', sep='\t')

# Agrupar los datos por 'ID' para visualización
grouped = data.groupby('ID')

# Preparar el gráfico
plt.figure(figsize=(15, 15))

# Variable para manejar la posición en el eje Y
y_pos = 0
yticks = []
legend_labels = set()  # Usar un conjunto para evitar duplicados en la leyenda

# Colores únicos para cada ID
colors = list(mcolors.TABLEAU_COLORS)  # Usar colores de la paleta de Matplotlib
color_map = {seq_id: colors[i % len(colors)] for i, seq_id in enumerate(grouped.groups.keys())}

# Graficar dominios para cada proteína
for seq_id, group in grouped:
    for idx, row in group.iterrows():
        # Añadir el tipo de dominio a la leyenda solo si no ha sido agregado previamente
        label = row['motif'] if row['motif'] not in legend_labels else None
        plt.barh(y_pos, row['End'] - row['Start'], left=row['Start'], color=color_map[seq_id], label=label, height=0.6)
        if label:
            legend_labels.add(row['motif'])  # Marcar este dominio como agregado a la leyenda

    # Agregar el nombre de la proteína a la lista de etiquetas del eje Y
    yticks.append(seq_id)
    y_pos += 1

# Configurar el gráfico
plt.xlabel('Posición en el promotor')
plt.ylabel('Promotor (ID)')
plt.yticks(range(len(yticks)), yticks)  # Asignar los nombres de proteínas al eje Y
plt.title('Distribución de dominios en las proteínas')
plt.grid(True, axis='x')

# Mostrar la leyenda sin duplicados
plt.legend(title='Motivos', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Guardar el gráfico como archivo SVG
plt.savefig('distribucion_dominios_promotores.svg', format='svg')
# Mostrar el gráfico
plt.show()
```

```python
# Leer los datos en un DataFrame
data = pd.read_csv('Ghiz_plot_Data.txt', sep='\t')

# Agrupar los datos por 'ID' para visualización
grouped = data.groupby('ID')

# Preparar el gráfico
plt.figure(figsize=(15, 10))

# Variable para manejar la posición en el eje Y
y_pos = 0
yticks = []
legend_labels = set()  # Usar un conjunto para evitar duplicados en la leyenda

# Colores únicos para cada ID
colors = list(mcolors.TABLEAU_COLORS)  # Usar colores de la paleta de Matplotlib
color_map = {seq_id: colors[i % len(colors)] for i, seq_id in enumerate(grouped.groups.keys())}

# Graficar dominios para cada proteína
for seq_id, group in grouped:
    for idx, row in group.iterrows():
        # Usar el color asignado al ID
        color = color_map[seq_id]
        
        # Añadir el tipo de dominio a la leyenda solo si no ha sido agregado previamente
        label = row['motif'] if row['motif'] not in legend_labels else None
        plt.plot([row['Start'], row['End']], [y_pos, y_pos], marker='o', linestyle='-', color=color, label=label)
        if label:
            legend_labels.add(row['motif'])  # Marcar este dominio como agregado a la leyenda

    # Agregar el nombre de la proteína a la lista de etiquetas del eje Y
    yticks.append(seq_id)
    y_pos += 1

# Configurar el gráfico
plt.xlabel('Posición en la proteína')
plt.ylabel('Proteína (ID)')
plt.yticks(range(len(yticks)), yticks)  # Asignar los nombres de proteínas al eje Y
plt.title('Distribución de dominios en las proteínas')
plt.grid(True, axis='x')

# Mostrar la leyenda sin duplicados
plt.legend(title='Motivos', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Guardar el gráfico como archivo SVG
plt.savefig('distribucion_dominios_promotores.svg', format='svg')

# Mostrar el gráfico
plt.show()
```

The following code was developed to get the genomic coordinates of the promoters regulating genes of the DAPseq experiment. 

```python
# Import libraries 
import pandas as pd
```

```python
# Define the function to get the promoters
def promotores_ran(df, processed_genes):
    promotores = []
    
    for i, fila in df.iterrows():
         # Get information about the gene 
        id_gen = fila['proteinId']
        
        # Check that it hasn't been processed yet
        if id_gen in processed_genes:
            continue
        start_gen = fila['start']
        end_gen = fila['end']
        scaffold_gen = fila['scaffold']
        strand = fila['strand']
        ID_promoter = "seq_" + str(id_gen)  # Convert id_gen to a string
        
        if strand == "+":
            startP = start_gen - 501
            endP = start_gen - 1
            longitudP = endP - startP
           
            if (startP < endP and startP> 0 and endP> 0) :  # Ensure that startP is less than endP
                promotores.append((id_gen, scaffold_gen, startP, endP, strand, ID_promoter, longitudP))
        elif strand == "-":
            startP = end_gen + 1
            endP = end_gen + 501
            longitudP = endP - startP
          
            if (startP < endP and startP> 0 and endP> 0) :  # Ensure that startP is less than endP
                promotores.append((id_gen, scaffold_gen, startP, endP, strand, ID_promoter, longitudP))
            
            # Mark the gene as processed
        processed_genes.add(id_gen)
            
    return promotores
```

```python
# Fijar la semilla y tomar una muestra aleatoria
seed = 123
# Depuración inicial de archivos
gff = pd.read_csv("Rhimi59_gff.csv", sep=",")
candidates = pd.read_csv("ID_List.txt", sep=",", header=None)

# Establecer la columna 'Gen_ID' como nombres de las columnas
candidates.columns = ['Gen_ID']

```

```python
# Realizar el merge
merge_result = pd.merge(candidates, gff, left_on="Gen_ID", right_on="proteinId", how="left", indicator=True)
merge_result
```

```python
# Convert 'proteinId' to an integer.
merge_result["proteinId"] = merge_result["proteinId"].astype(int)

# Call the function for the gene data
processed_genes = set()  # Initialize the set of processed genes
promotores = promotores_ran(merge_result, processed_genes)
print(promotores)

# Verify if promoters were found
if len(promotores) == 0:
    print("No promoters were found.")
else:
    # Convert the list of promoters to a DataFrame
    df_promotores = pd.DataFrame(promotores, columns=["proteinID", 'scaffold', 'start', 'end', 'strand', 'ID', 'longitud'])
    
    # Filter the desired columns
    df_promotores_sample = df_promotores[[ 'scaffold', 'start', 'end',"proteinID"]]
    df_promotores_sample["start"]= df_promotores_sample["start"].astype(int)
    df_promotores_sample["end"]=df_promotores_sample["end"].astype(int)
    
    # Save to a text file
    nombre_salida_sample = "promotores_DAP_sample_promoters.txt"
    df_promotores_sample.to_csv(nombre_salida_sample, index=False, sep="\t", header=False)

    print(f"File saved as {nombre_salida_sample}")
   
```
